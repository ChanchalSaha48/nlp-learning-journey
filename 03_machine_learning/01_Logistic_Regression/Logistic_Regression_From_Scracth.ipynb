```python
import numpy as np
```

```python
# Dataset
X=np.array([1,2,3,4,5,6],dtype=float)
y=np.array([0,0,0,1,1,1],dtype=float)
```

```python
print(f"X={X}")
print(f"y={y}")
```

```python
# Sigmoid function

def sigmoid(z):
  return 1/(1+np.exp(-z))
```

```python
# Initial function
w=0.0   # w--> weight
b=0.0   # b--> bias

learning_rate=0.1
epochs=100

loss_history=[]
weight_history=[]
bias_history=[]
```

```python
for epoch in range(epochs):
  # Forward pass
  z=(w*X)+b
  p=sigmoid(z) # p--> probability

  # Loss
  loss=-np.mean(
      (y*np.log(p + 1e-15)) +((1-y)*np.log(1 - p + 1e-15))
      )

  # Gradients
  dw=np.mean((p-y)*X)
  db=np.mean(p-y)

  # Updates weights
  w=w-learning_rate*dw
  b=b-learning_rate*db

  print(f"Epoch{epoch+1:2d} | "
        f"Loss:{loss:.4f} | "
        f"Weight: {w:.4f} | ",
        f"Bias: {b:.4f}")
  # Save history
  loss_history.append(loss)
  weight_history.append(w)
  bias_history.append(b)
```

```python
# Graph

import matplotlib.pyplot as plt
```

```python
plt.plot(loss_history)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Loss During Training")
plt.show()
```

```python
plt.plot(weight_history)
plt.xlabel("Epoch")
plt.ylabel("Weight")
plt.title("Weight During Training")
plt.show()
```

```python
plt.plot(bias_history)
plt.xlabel("Epoch")
plt.ylabel("Bias")
plt.title("Bias During Training")
plt.show()
```

```python
print(f"Initial loss{loss_history[0]}")
print(f"Final loss{loss_history[-1]}")

print(f"Initial weight{weight_history[0]}")
print(f"final weight{weight_history[-1]}")
```
